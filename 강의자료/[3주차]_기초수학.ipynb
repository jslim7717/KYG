```python
def factorial(x):
    """
    팩토리얼 함수
    입력값: 정수 x
    출력값: x!
    """
    x_list = list(range(1, x+1))
    res = 1
    for val in x_list:
        res *= val
    return res

def combination(n, x):
    """
    조합
    입력값: n, x
    출력값: nCx(실수)
    """
    res = factorial(n)/(factorial(x)*factorial(n-x))
    return res
```

```python
factorial(5)
```

```python
combination(5, 3)
```

# 자연 상수 e

```python
def factorial(x):
    """
    팩토리얼 함수
    입력값: 정수 x
    출력값: x!
    """
    x_list = list(range(1, x+1))
    res = 1
    for val in x_list:
        res *= val
    return res
```

```python
n = 10
e = 0
for i in range(0,n):
    e += 1/factorial(i)
print(e)
```

# 수열

```python
def seq(start, stop, step):
    """
    수열 만들기
    입력값: start(시작 값), stop(끝 값), step(한 스텝당 증가 수)
    출력값: res(리스트)
    """
    res = []
    current = start
    while current < stop:
        res.append(current)
        current += step
    return res
```

```python
seq(2,10,3)
```

# 로그

```python
def log(x):
    """
    밑이 e인 자연 로그
    입력값: 진수 x
    출력값: ln(x)
    """
    n = 100000000.0
    res = n*( ( x**(1/n) ) -1 ) #로피탈 정리
    return res
```

# Gradient

```python
import numpy as np
```

```python
x = np.array([2, 4, 6, 8])
y = np.array([81, 93, 91, 97])
print(x)
```

```python
mx = np.mean(x)
my = np.mean(y)
print("x 평균:", mx)
print("y 평균:", my)
```

```python
divisor = sum([(i - mx)**2 for i in x])
print(divisor)

def top(x, mx, y, my):
    d = 0
    for i in range(len(x)):
        d += (x[i] - mx) * (y[i] - my)
    return d
dividend = top(x, mx, y, my)
print("분모:", divisor)
print("분자:", dividend)
```

```python
w = dividend / divisor
b = my - (mx*w)

print("기울기 =", w)
print("y절편 =", b)
```

## ML: Gradient

```python
# 기울기 w와 절편 b의 값을 초기화합니다.
w = 0
b = 0

# 학습률을 정합니다.
learning_rate = 0.03

# 몇 번 반복될지 설정합니다.
epochs = 2000

# x 값이 총 몇 개인지 셉니다.
n=len(x)

# 경사 하강법을 시작합니다.
for i in range(1,epochs):
    y_pred = w * x + b
    error = y - y_pred

    w_diff = (2/n) * sum(-x * (error))
    b_diff = (2/n) * sum(-(error))

    w = w - learning_rate * w_diff
    b = b - learning_rate * b_diff

    if i % 100 == 0:
        print("epoch=%.f, 기울기=%.4f,절편=%.4f" % (i, w, b))
```
